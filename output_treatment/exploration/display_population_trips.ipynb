### Imports utiles

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

from IPython.display import HTML
```

### Tableaux comparatifs variables socio-économiques (pop synthétique vs. théorique)

##### Nombre de déplacements par trage x purpose

```python
# Charger le CSV
df_age_purpose = pd.read_csv("output/analysis_population/simulated_age_purpose.csv")

# Transformation longue pour Plotly
df_age_purpose_melted = df_age_purpose.melt(id_vars="age_class", var_name="Trip Purpose", value_name="Count")

# Tracé
fig = px.bar(
    df_age_purpose_melted,
    x="age_class",
    y="Count",
    color="Trip Purpose",
    barmode="group",
    title="Répartition des trajets par classe d'âge et type de déplacement",
    labels={"age_class": "Classe d'âge", "Count": "Nombre de trajets"}
)

fig.show()
```

##### Effectifs par trage

```python
# Charger les données
df = pd.read_csv("output/analysis_population/simulated_age.csv")

# Initialiser la figure
fig = go.Figure()

# Ajouter les barres INSEE
fig.add_trace(go.Bar(
    x=df["Age class"],
    y=df["INSEE"],
    name="INSEE",
    text=(df["Proportion_INSEE"] * 100).round(1).astype(str) + '%',
    textposition='outside',
    marker_color='royalblue'
))

# Ajouter les barres EQASIM
fig.add_trace(go.Bar(
    x=df["Age class"],
    y=df["EQASIM"],
    name="EQASIM",
    text=(df["Proportion_EQASIM"] * 100).round(1).astype(str) + '%',
    textposition='outside',
    marker_color='indianred'
))

# Configuration de la mise en page
fig.update_layout(
    barmode='group',
    title="Effectifs par classe d'âge : INSEE vs EQASIM",
    xaxis_title="Classe d'âge",
    yaxis_title="Effectif",
    legend_title="Source",
    uniformtext_minsize=8,
    uniformtext_mode='hide'
)

fig.show()
```

##### Nombre de déplacements par distance

```python
# Charger les données
df = pd.read_csv("output/analysis_population/simulated_distance.csv")

# On garde uniquement les distances avec données > 0 (facultatif si beaucoup de 0)
# df = df[(df["HTS adjust with INSEE population weights"] > 0) | (df["EQASIM"] > 0)]

# Création de la figure
fig = go.Figure()

# Barres HTS ajustées
fig.add_trace(go.Bar(
    x=df["Distance class"],
    y=df["HTS adjust with INSEE population weights"],
    name="HTS (pondéré)",
    text=(df["Proportion_HTS"] * 100).round(2).astype(str) + '%',
    textposition="outside",
    marker_color='steelblue'
))

# Barres EQASIM
fig.add_trace(go.Bar(
    x=df["Distance class"],
    y=df["EQASIM"],
    name="EQASIM",
    text=(df["Proportion_EQASIM"] * 100).round(2).astype(str) + '%',
    textposition="outside",
    marker_color='tomato'
))

# Mise en page
fig.update_layout(
    barmode="group",
    title="Comparaison des déplacements par classe de distance : HTS vs EQASIM",
    xaxis_title="Classe de distance (km)",
    yaxis_title="Nombre de déplacements",
    xaxis_tickangle=45,
    legend_title="Source",
    uniformtext_minsize=8,
    uniformtext_mode='hide',
    height=600
)

fig.show()
```

##### Effectifs selon la possession du permis de conduire

```python
# Charger les données
df = pd.read_csv("output/analysis_population/simulated_license.csv")

# Convertir les valeurs booléennes en libellés clairs
df["Possession of license"] = df["Possession of license"].map({True: "Oui", False: "Non"})

# Créer la figure
fig = go.Figure()

# Barres HTS ajusté
fig.add_trace(go.Bar(
    x=df["Possession of license"],
    y=df["HTS adjust with INSEE population weights"],
    name="HTS (pondéré)",
    text=(df["Proportion_HTS"] * 100).round(1).astype(str) + '%',
    textposition='outside',
    marker_color='royalblue'
))

# Barres EQASIM
fig.add_trace(go.Bar(
    x=df["Possession of license"],
    y=df["EQASIM"],
    name="EQASIM",
    text=(df["Proportion_EQASIM"] * 100).round(1).astype(str) + '%',
    textposition='outside',
    marker_color='firebrick'
))

# Mise en page
fig.update_layout(
    barmode='group',
    title="Comparaison de la possession du permis : HTS vs EQASIM",
    xaxis_title="Possession du permis de conduire",
    yaxis_title="Nombre de personnes",
    legend_title="Source",
    uniformtext_minsize=8,
    uniformtext_mode='hide'
)

fig.show()
```

##### Effectifs selon le nombre de véhicules

```python
# Charger les données
df = pd.read_csv("C:/VSCodeProjects/matsim-idf/output/analysis_population/simulated_nbr_vehicle.csv")

# Convertir les classes en libellés texte plus explicites (facultatif)
df["Number of vehicles class"] = df["Number of vehicles class"].astype(str)
df["Number of vehicles class"] = df["Number of vehicles class"].replace({"3+": "3 et plus"})

# Créer la figure
fig = go.Figure()

# Barres HTS ajusté
fig.add_trace(go.Bar(
    x=df["Number of vehicles class"],
    y=df["HTS"],
    name="HTS",
    text=(df["Proportion_HTS"] * 100).round(1).astype(str) + '%',
    textposition='outside',
    marker_color='dodgerblue'
))

# Barres EQASIM
fig.add_trace(go.Bar(
    x=df["Number of vehicles class"],
    y=df["EQASIM"],
    name="EQASIM",
    text=(df["Proportion_EQASIM"] * 100).round(1).astype(str) + '%',
    textposition='outside',
    marker_color='crimson'
))

# Mise en page
fig.update_layout(
    barmode='group',
    title="Nombre de véhicules par ménage : comparaison HTS vs EQASIM",
    xaxis_title="Nombre de véhicules",
    yaxis_title="Nombre de ménages",
    legend_title="Source",
    uniformtext_minsize=8,
    uniformtext_mode='hide'
)

fig.show()
```
